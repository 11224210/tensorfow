```python
import glob

# 尋找並修補 tensorflow_io 的載入問題
files = glob.glob('/content/models/research/object_detection/**/*.py', recursive=True)
count = 0
for f in files:
    try:
        with open(f, 'r', encoding='utf-8') as file:
            content = file.read()

        if 'import tensorflow_io as tfio' in content:
            # 將原本的 import 換成 try-except 來略過報錯
            content = content.replace('import tensorflow_io as tfio',
                                      'try:\n  import tensorflow_io as tfio\nexcept ImportError:\n  tfio = None')
            with open(f, 'w', encoding='utf-8') as file:
                file.write(content)
            count += 1
    except:
        pass

print(f"✅ 成功略過 tensorflow_io，修補了 {count} 個檔案！")
```

```python
import glob
import os

# 1. 復原剛才破壞的縮排結構
files = glob.glob('/content/models/research/object_detection/**/*.py', recursive=True)
for f in files:
    try:
        with open(f, 'r', encoding='utf-8') as file:
            content = file.read()

        # 將錯誤的替換復原回原本的 import
        bad_code = 'try:\n  import tensorflow_io as tfio\nexcept ImportError:\n  tfio = None'
        if bad_code in content:
            content = content.replace(bad_code, 'import tensorflow_io as tfio')
            with open(f, 'w', encoding='utf-8') as file:
                file.write(content)
    except:
        pass

# 2. 建立空殼的 tensorflow_io 套件
os.makedirs('/usr/local/lib/python3.13/dist-packages/tensorflow_io', exist_ok=True)
with open('/usr/local/lib/python3.13/dist-packages/tensorflow_io/__init__.py', 'w') as f:
    f.write('# 空殼模組')

print("✅ 縮排錯誤已修復，空殼 tensorflow_io 建立完成！")
```

```python
import os
import io
import glob
import pandas as pd
import xml.etree.ElementTree as ET
import tensorflow as tf
from PIL import Image
from object_detection.utils import dataset_util
from collections import namedtuple

# 設定絕對路徑
base_path = '/content/Tensorflow-Object-Detection-API-With-Custom-Dataset'
data_path = os.path.join(base_path, 'data')
os.makedirs(data_path, exist_ok=True)

# 1. ================== XML 轉 CSV ==================
def xml_to_csv(path):
    xml_list = []
    for xml_file in glob.glob(path + '/*.xml'):
        tree = ET.parse(xml_file)
        root = tree.getroot()
        for member in root.findall('object'):
            value = (root.find('filename').text, int(root.find('size')[0].text),
                     int(root.find('size')[1].text), member[0].text,
                     int(member.find('bndbox')[0].text), int(member.find('bndbox')[1].text),
                     int(member.find('bndbox')[2].text), int(member.find('bndbox')[3].text))
            xml_list.append(value)
    return pd.DataFrame(xml_list, columns=['filename', 'width', 'height', 'class', 'xmin', 'ymin', 'xmax', 'ymax'])

for directory in ['train', 'test']:
    img_path = os.path.join(base_path, 'images', directory)
    if os.path.exists(img_path) and len(glob.glob(img_path + '/*.xml')) > 0:
        df = xml_to_csv(img_path)
        df.to_csv(os.path.join(data_path, f'{directory}_labels.csv'), index=None)
        print(f'✅ CSV 處理完成：{directory}_labels.csv')
    else:
        print(f'❌ 警告：在 {directory} 資料夾找不到 XML，請確認圖片已上傳！')

# 2. ================== CSV 轉 TFRecord ==================
def class_text_to_int(row_label):
    if row_label == 'apple': return 1
    else: return 1  # 防呆機制，只要有框預設都給 ID 1

def split(df, group):
    data = namedtuple('data', ['filename', 'object'])
    gb = df.groupby(group)
    return [data(filename, gb.get_group(x)) for filename, x in zip(gb.groups.keys(), gb.groups)]

for directory in ['train', 'test']:
    csv_path = os.path.join(data_path, f'{directory}_labels.csv')
    img_path = os.path.join(base_path, 'images', directory)
    record_path = os.path.join(data_path, f'{directory}.record')

    if os.path.exists(csv_path):
        writer = tf.io.TFRecordWriter(record_path)
        examples = pd.read_csv(csv_path)
        grouped = split(examples, 'filename')
        for group in grouped:
            with tf.io.gfile.GFile(os.path.join(img_path, group.filename), 'rb') as fid:
                encoded_jpg = fid.read()
            encoded_jpg_io = io.BytesIO(encoded_jpg)
            image = Image.open(encoded_jpg_io)
            width, height = image.size

            xmins, xmaxs, ymins, ymaxs, classes_text, classes = [], [], [], [], [], []
            for index, row in group.object.iterrows():
                xmins.append(row['xmin'] / width)
                xmaxs.append(row['xmax'] / width)
                ymins.append(row['ymin'] / height)
                ymaxs.append(row['ymax'] / height)
                classes_text.append(row['class'].encode('utf8'))
                classes.append(class_text_to_int(row['class']))

            tf_example = tf.train.Example(features=tf.train.Features(feature={
                'image/height': dataset_util.int64_feature(height),
                'image/width': dataset_util.int64_feature(width),
                'image/filename': dataset_util.bytes_feature(group.filename.encode('utf8')),
                'image/source_id': dataset_util.bytes_feature(group.filename.encode('utf8')),
                'image/encoded': dataset_util.bytes_feature(encoded_jpg),
                'image/format': dataset_util.bytes_feature(b'jpg'),
                'image/object/bbox/xmin': dataset_util.float_list_feature(xmins),
                'image/object/bbox/xmax': dataset_util.float_list_feature(xmaxs),
                'image/object/bbox/ymin': dataset_util.float_list_feature(ymins),
                'image/object/bbox/ymax': dataset_util.float_list_feature(ymaxs),
                'image/object/class/text': dataset_util.bytes_list_feature(classes_text),
                'image/object/class/label': dataset_util.int64_list_feature(classes),
            }))
            writer.write(tf_example.SerializeToString())
        writer.close()
        print(f'🔥 TFRecord 建立大成功：{record_path}')
```
